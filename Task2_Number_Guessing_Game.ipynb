# Task 2: Random Number Guessing Game

This program generates a random number and asks the user to guess it.

```python
import random

print("🎯 Welcome to Number Guessing Game")

number = random.randint(1, 100)
attempts = 0

while True:
    guess = int(input("Enter your guess (1-100): "))
    attempts += 1

    if guess == number:
        print("🎉 Correct! You guessed the number.")
        print("Total attempts:", attempts)
        break
    elif guess > number:
        print("📉 Too high! Try again.")
    else:
        print("📈 Too low! Try again.")
```

## Explanation

- The program generates a random number between 1 and 100.
- The user guesses the number.
- The program gives hints (too high or too low).
- It repeats until the correct number is guessed.
- It also counts the number of attempts.
